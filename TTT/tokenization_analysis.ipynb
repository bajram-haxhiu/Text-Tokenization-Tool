```python
import pandas as pd
import spacy
from nltk.tokenize import word_tokenize

nlp = spacy.blank("en")

sentences = {
    "Contractions": "I can't believe we're learning NLP!",
    "Numbers": "Revenue rose from 1,234.56 to 1,450.20 dollars.",
    "Dates": "Call me at 09:15 on 2025-01-31.",
    "Punctuation": "Hello!!! Are you ready? Yes, I am.",
    "Email and URL": "Contact me at test@example.com or visit https://example.com."
}

results = []

for category, text in sentences.items():
    split_tokens = text.split()
    nltk_tokens = word_tokenize(text)
    spacy_tokens = [token.text for token in nlp(text)]

    results.append({
        "Category": category,
        "Characters": len(text),
        "split() tokens": len(split_tokens),
        "NLTK tokens": len(nltk_tokens),
        "spaCy tokens": len(spacy_tokens)
    })

df = pd.DataFrame(results)
df
```

```python
import matplotlib.pyplot as plt

df.plot(
    x="Category",
    y=["split() tokens", "NLTK tokens", "spaCy tokens"],
    kind="bar",
    figsize=(10, 5)
)

plt.title("Tokenization Methods Comparison")
plt.ylabel("Number of Tokens")
plt.xlabel("Text Category")
plt.xticks(rotation=25, ha="right")
plt.tight_layout()
plt.show()
```

```python
from collections import Counter
import pandas as pd
import matplotlib.pyplot as plt

text = """
Natural Language Processing (NLP) helps computers understand human language.
NLP is used in chatbots, translation, search engines, and text analysis.
Python is a popular language for NLP because it has powerful libraries.
"""

# Tokenization
split_tokens = text.split()
nltk_tokens = word_tokenize(text)
spacy_tokens = [token.text for token in nlp(text)]

# Basic statistics
statistics = {
    "Metric": [
        "Characters",
        "Words using split()",
        "Tokens using NLTK",
        "Tokens using spaCy",
        "Unique words using split()"
    ],
    "Value": [
        len(text),
        len(split_tokens),
        len(nltk_tokens),
        len(spacy_tokens),
        len(set(word.lower() for word in split_tokens))
    ]
}

stats_df = pd.DataFrame(statistics)
stats_df
```

```python
import re
from collections import Counter

words = re.findall(r"\b[a-zA-Z]+\b", text.lower())
word_counts = Counter(words)

most_common = pd.DataFrame(
    word_counts.most_common(10),
    columns=["Word", "Frequency"]
)

most_common
```

```python
most_common.plot(
    x="Word",
    y="Frequency",
    kind="bar",
    legend=False,
    figsize=(9, 4)
)

plt.title("Top 10 Most Frequent Words")
plt.xlabel("Word")
plt.ylabel("Frequency")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
